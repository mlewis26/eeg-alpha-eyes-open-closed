Simulated data: 8 made up records, used to build and test the analysis

1) What is the average fever for simple seizures?
2) What is the highest fever among simple seizures?
3) What is the average age (in months) for simple seizures?
4) How many seizures in the log are complex (longer than 15 minutes)?
5) How does the average fever compare between simple and complex seizures?

```python
seizures = [{"patient": "P01", "age_months": 6, "fever_c": 39.2, "minutes": 4},
            {"patient": "P02", "age_months": 27, "fever_c": 41.1, "minutes": 7},
            {"patient": "P03", "age_months": 40, "fever_c": 40.0, "minutes": 20},
            {"patient": "P04", "age_months": 13, "fever_c": 38.5, "minutes": 10},
            {"patient": "P05", "age_months": 26, "fever_c": 39.3, "minutes": 9},
            {"patient": "P06", "age_months": 58, "fever_c": 41.0, "minutes": 16},
            {"patient": "P07", "age_months": 7, "fever_c": 38.9, "minutes": 18},
            {"patient": "P08", "age_months": 11, "fever_c": 40.7, "minutes": 9}]
```

Q1: What is the average fever for simple seizures?

```python
simple_fevers = []
for s in seizures:
  if s["minutes"] <= 15:
    simple_fevers.append(s["fever_c"])
print(simple_fevers)
print(round(sum(simple_fevers) / len(simple_fevers), 2))
```

Q2: What is the highest temperature among simple seizures?

```python
print(max(simple_fevers))
```

Q3: What is the average age in months for simple seizures?

```python
simple_ages = []
for s in seizures:
  if s["minutes"] <= 15:
    simple_ages.append(s["age_months"])
print(simple_ages)
print(sum(simple_ages) / len(simple_ages))
```

Q4: How many seizures in the log are complex (longer than 15 minutes)?

```python
complex_seizures = []
for s in seizures:
  if s["minutes"] > 15:
    complex_seizures.append(s["patient"])
print(complex_seizures)
print(len(complex_seizures))
```

Q5: How does the average fever compare between simple and complex seizures?

```python
simple_fevers = []
for s in seizures:
  if s["minutes"] <= 15:
    simple_fevers.append(s["fever_c"])
print(simple_fevers)
print(round(sum(simple_fevers) / len(simple_fevers), 2))

complex_fevers = []
for s in seizures:
  if s["minutes"] > 15:
    complex_fevers.append(s["fever_c"])
print(complex_fevers)
print(round(sum(complex_fevers) / len(complex_fevers), 2))
```
